# Python Data Analysis — Sales Dataset

A complete Python-based sales data analysis workflow using a synthetic dataset of 300 sales transactions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Python Data Analysis Environment: READY")
```

```python
rng = np.random.default_rng(42)

n = 300

sales_df = pd.DataFrame({
    "Product": rng.choice(["Laptop", "Phone", "Tablet", "Monitor", "Keyboard"], size=n),
    "Region": rng.choice(["North", "South", "East", "West"], size=n),
    "Units_Sold": rng.integers(1, 20, size=n),
    "Unit_Price": np.round(rng.uniform(20, 1000, size=n), 2)
})

sales_df["Revenue"] = (sales_df["Units_Sold"] * sales_df["Unit_Price"]).round(2)

print("Dataset created successfully.")
print("Rows   :", len(sales_df))
print("Columns:", len(sales_df.columns))

display(sales_df.head())
```

```python
print("=" * 60)
print("DATA QUALITY AUDIT")
print("=" * 60)

print("\n1. Dataset shape:")
print(sales_df.shape)

print("\n2. Column data types:")
print(sales_df.dtypes)

print("\n3. Missing values:")
print(sales_df.isnull().sum())

print("\n4. Duplicate rows:")
print(sales_df.duplicated().sum())

print("\n5. Basic statistics:")
display(sales_df.describe())

print("\n6. Unique values per categorical column:")
print("Product:", sales_df["Product"].nunique())
print("Region :", sales_df["Region"].nunique())

print("\nDATA QUALITY AUDIT: COMPLETE")
```

```python
# Product-wise Revenue Analysis
product_revenue = (
    sales_df.groupby("Product", as_index=False)["Revenue"]
    .sum()
    .sort_values("Revenue", ascending=False)
)

print("Product-wise Revenue:")
display(product_revenue)

print("\nHighest-revenue product:")
top_product = product_revenue.iloc[0]
print(f"{top_product['Product']} → Revenue = {top_product['Revenue']:,.2f}")
```

```python
# Revenue Visualization
plt.figure(figsize=(9, 5))
plt.bar(product_revenue["Product"], product_revenue["Revenue"])
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Total Revenue")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Region-wise Revenue Analysis
region_revenue = (
    sales_df.groupby("Region", as_index=False)["Revenue"]
    .sum()
    .sort_values("Revenue", ascending=False)
)

print("Region-wise Revenue:")
display(region_revenue)

top_region = region_revenue.iloc[0]
print("\nHighest-revenue region:")
print(f"{top_region['Region']} → Revenue = {top_region['Revenue']:,.2f}")
```

```python
# Product × Region Revenue Analysis
product_region_revenue = (
    sales_df.pivot_table(index="Product", columns="Region", values="Revenue", aggfunc="sum")
    .round(2)
)

print("Product × Region Revenue:")
display(product_region_revenue)

pair_series = product_region_revenue.stack().sort_values(ascending=False)
best_pair = pair_series.iloc[0]
best_product, best_region = pair_series.index[0]

print("\nHighest product-region combination:")
print(f"{best_product} in {best_region} → Revenue = {best_pair:,.2f}")
```

```python
# Final Business Insights
total_revenue = sales_df["Revenue"].sum()
total_units = sales_df["Units_Sold"].sum()

print("=" * 60)
print("FINAL BUSINESS INSIGHTS")
print("=" * 60)

print(f"\nTotal Revenue       : {total_revenue:,.2f}")
print(f"Total Units Sold    : {total_units:,}")
print(f"\nTop Product         : {top_product["Product"]} → {top_product["Revenue"]:,.2f}")
print(f"Top Region          : {top_region["Region"]} → {top_region["Revenue"]:,.2f}")
print(f"\nTop Product-Region  : {best_product} + {best_region} → {best_pair:,.2f}")
print("\n" + "=" * 60)
print("ANALYSIS COMPLETE")
print("=" * 60)
```

```python
# Portfolio KPI Summary
kpi_summary = pd.DataFrame({
    "KPI": ["Total Revenue", "Total Units Sold", "Top Product", "Top Region", "Top Product × Region"],
    "Result": [
        f"{total_revenue:,.2f}",
        f"{total_units:,}",
        f"{top_product['Product']} ({top_product['Revenue']:,.2f})",
        f"{top_region['Region']} ({top_region['Revenue']:,.2f})",
        f"{best_product} + {best_region} ({best_pair:,.2f})"
    ]
})

display(kpi_summary)
```
